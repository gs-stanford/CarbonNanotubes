# Carbon Property Tables: quickstart
Create a citation-backed comparison figure with readable property names. The service returns a rendered figure, not the complete canonical coordinate table.

```python
from pathlib import Path

import carbon_property_tables as cpt

print(f"Carbon Property Tables SDK {cpt.__version__}")
```

```python
catalog = cpt.properties()
print(f"{len(catalog)} supported properties")
[(item["label"], item["display_unit"]) for item in catalog]
```

```python
scatter = cpt.scatter(
    "specific strength",
    "specific conductivity",
    log_x=True,
    log_y=True,
    peer_reviewed=True,
    top=5,
    top_by="y",
)
scatter
```

```python
output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)
figure_paths = scatter.save_bundle(output_dir / "specific-conductivity-vs-specific-strength")
table_path = scatter.save_top_table(output_dir / "top-five.csv")
print(figure_paths)
print(table_path)
```

```python
for row in scatter.top_table():
    print(row["rank"], row["label"], row["y_value"], row["y_unit"], row["doi"])
```
