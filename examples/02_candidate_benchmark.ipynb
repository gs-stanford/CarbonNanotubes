# Benchmark a temporary CNT result
Place an unpublished or newly measured point against the visible peer-reviewed CNT landscape. Temporary points are rendered and ranked but are never written to Carbon Property Tables.

```python
from pathlib import Path

import carbon_property_tables as cpt
from carbon_property_tables import TemporaryPoint
```

```python
candidate = TemporaryPoint(
    x=1.8,  # N tex^-1: display unit on the x-axis
    y=12.0,  # kS m^2 kg^-1: display unit on the y-axis
    label="My candidate CNT fiber",
)

comparison = cpt.scatter(
    "specific strength",
    "specific cond",
    log_x=True,
    log_y=True,
    peer_reviewed=True,
    material_family=["CNT_or_CNT_hybrid", "CNT_metal_composite"],
    top=5,
    top_by="y",
    formats=("svg", "pdf"),
    temporary=candidate,
)
comparison
```

```python
rank = comparison.temporary_point
print(f"x rank: {rank.x_rank}/{rank.total_with_temporary}")
print(f"y rank: {rank.y_rank}/{rank.total_with_temporary}")
print(f"Pareto frontier: {rank.on_pareto_frontier}")
```

```python
output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)
comparison.save(output_dir / "candidate-comparison.svg")
comparison.save(output_dir / "candidate-comparison.pdf")
comparison.save_top_table(output_dir / "candidate-top-five.csv")
sorted(path.name for path in output_dir.iterdir())
```
